```python
import pandas as pd
import numpy as np
```

```python
COLS_TO_REMOVE = [
    # слабо связаны с ценой
    'WoodDeckSF', '2ndFlrSF', 'OpenPorchSF', 'HalfBath',
    'LotArea', 'BsmtFullBath', 'BsmtUnfSF', 'BedroomAbvGr', 'ScreenPorch',
    'PoolArea', 'MoSold', '3SsnPorch', 'BsmtFinSF2', 'BsmtHalfBath',
    'MiscVal', 'Id', 'LowQualFinSF', 'YrSold', 'OverallCond', 'MSSubClass',
    'EnclosedPorch', 'KitchenAbvGr',
    # дублируют другие признаки
    '1stFlrSF', 'TotRmsAbvGrd', 'GarageCars',
    # исходная цена: вместо неё будет LogSalePrice
    'SalePrice',
    # пропущено больше 70% значений
    'PoolQC', 'MiscFeature', 'Alley', 'Fence',
]

COLS_TO_REMOVE_TEST = [
    # слабо связаны с ценой (Id в test оставляем: он нужен для сабмита)
    'WoodDeckSF', '2ndFlrSF', 'OpenPorchSF', 'HalfBath',
    'LotArea', 'BsmtFullBath', 'BsmtUnfSF', 'BedroomAbvGr', 'ScreenPorch',
    'PoolArea', 'MoSold', '3SsnPorch', 'BsmtFinSF2', 'BsmtHalfBath',
    'MiscVal', 'LowQualFinSF', 'YrSold', 'OverallCond', 'MSSubClass',
    'EnclosedPorch', 'KitchenAbvGr',
    # дублируют другие признаки
    '1stFlrSF', 'TotRmsAbvGrd', 'GarageCars',
    # пропущено больше 70% значений
    'PoolQC', 'MiscFeature', 'Alley', 'Fence',
]
```

```python
data = pd.read_csv('../data/raw/train.csv')
data['LogSalePrice'] = np.log1p(data['SalePrice'])
print(data.shape)
data.head()
```

```python
test_data = pd.read_csv('../data/raw/test.csv')
print(test_data.shape)
test_data.head()
```

```python
data.drop(columns=COLS_TO_REMOVE, inplace=True)
print(data.shape)
data.info()
```

```python
missing = data.isna().sum().index
missing
```

```python
test_data.drop(columns=COLS_TO_REMOVE_TEST, inplace=True)
print(test_data.shape)
test_data.info()
```

```python
NONE_COLS = ['MasVnrType', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
             'FireplaceQu', 'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond']
ZERO_COLS = ['MasVnrArea', 'BsmtFinSF1', 'TotalBsmtSF', 'GarageArea']

for df in (data, test_data):
    df[NONE_COLS] = df[NONE_COLS].fillna('None')
    df[ZERO_COLS] = df[ZERO_COLS].fillna(0)
    # Нет гаража -> год постройки дома (0 был бы выбросом среди годов ~1900-2010)
    df['GarageYrBlt'] = df['GarageYrBlt'].fillna(df['YearBuilt'])
```

```python
# LotFrontage: медиана по району (у соседних домов похожая ширина участка)
lot_medians = data.groupby('Neighborhood')['LotFrontage'].median()
for df in (data, test_data):
    df['LotFrontage'] = df['LotFrontage'].fillna(df['Neighborhood'].map(lot_medians))

# Остальные категориальные (MSZoning, Electrical, KitchenQual, SaleType, ...): мода из train
cat_cols = data.select_dtypes(exclude='number').columns
modes = data[cat_cols].mode().iloc[0]
data[cat_cols] = data[cat_cols].fillna(modes)
test_data[cat_cols] = test_data[cat_cols].fillna(modes)

print('NaN в train:', data.isna().sum().sum(), '| NaN в test:', test_data.isna().sum().sum())
```

```python
# 'None' значит «нет объекта» (нет подвала, гаража), поэтому он получает 0
QUAL_MAP = {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5}
FIN_TYPE_MAP = {'None': 0, 'Unf': 1, 'LwQ': 2, 'Rec': 3, 'BLQ': 4, 'ALQ': 5, 'GLQ': 6}

ORDINAL_MAPS = {
    'ExterQual': QUAL_MAP,
    'ExterCond': QUAL_MAP,
    'BsmtQual': QUAL_MAP,
    'BsmtCond': QUAL_MAP,
    'HeatingQC': QUAL_MAP,
    'KitchenQual': QUAL_MAP,
    'FireplaceQu': QUAL_MAP,
    'GarageQual': QUAL_MAP,
    'GarageCond': QUAL_MAP,
    'BsmtExposure': {'None': 0, 'No': 1, 'Mn': 2, 'Av': 3, 'Gd': 4},
    'BsmtFinType1': FIN_TYPE_MAP,
    'BsmtFinType2': FIN_TYPE_MAP,
    'GarageFinish': {'None': 0, 'Unf': 1, 'RFn': 2, 'Fin': 3},
    'Functional': {'Sal': 0, 'Sev': 1, 'Maj2': 2, 'Maj1': 3, 'Mod': 4, 'Min2': 5, 'Min1': 6, 'Typ': 7},
    'CentralAir': {'N': 0, 'Y': 1},
    'PavedDrive': {'N': 0, 'P': 1, 'Y': 2},
}

# заменяем текст на числа по словарю
for df in (data, test_data):
    for col, mapping in ORDINAL_MAPS.items():
        df[col] = df[col].map(mapping)
```

```python
nominal_cols = data.select_dtypes(exclude='number').columns
print(f'Nominal features: {len(nominal_cols)}')

data = pd.get_dummies(data, columns=nominal_cols, dtype=int)
test_data = pd.get_dummies(test_data, columns=nominal_cols, dtype=int)

# В test могут не встретиться некоторые категории из train (и наоборот),
# поэтому выравниваем колонки test по train: недостающие заполняем нулями, лишние удаляем.
# LogSalePrice есть только в train, Id только в test
feature_cols = data.columns.drop('LogSalePrice')
test_data = test_data.reindex(columns=['Id', *feature_cols], fill_value=0)
```

```python
print(data.shape, test_data.shape)
print('NaN в train:', data.isna().sum().sum(), '| NaN в test:', test_data.isna().sum().sum())
print('Нечисловых колонок:', len(data.select_dtypes(exclude='number').columns))
```

```python
CONTINUOUS_COLS = ['LotFrontage', 'MasVnrArea', 'BsmtFinSF1', 'TotalBsmtSF', 'GrLivArea', 'GarageArea']

skew_before = data[CONTINUOUS_COLS].skew()
skew_after = np.log1p(data[CONTINUOUS_COLS]).skew()

# Логарифмируем, только если признак скошен и логарифм реально уменьшает скошенность.
# Например, у TotalBsmtSF есть нули (нет подвала): после log1p они становятся выбросами слева
log_cols = skew_before[(skew_before > 0.75) & (skew_after.abs() < skew_before.abs())].index

print(pd.DataFrame({'before': skew_before, 'after_log': skew_after}).round(2))
print('\nLog-transformed:', list(log_cols))

for df in (data, test_data):
    df[log_cols] = np.log1p(df[log_cols])
```

```python
# Масштабируем все небинарные признаки (площади, годы, ordinal-оценки); one-hot колонки 0/1 не трогаем
scale_cols = [col for col in feature_cols if data[col].nunique() > 2]
print(f'Scaled features: {len(scale_cols)}')

means = data[scale_cols].mean()
stds = data[scale_cols].std()

data[scale_cols] = (data[scale_cols] - means) / stds
test_data[scale_cols] = (test_data[scale_cols] - means) / stds

data[scale_cols].describe().loc[['mean', 'std']].round(2)
```

```python
import os

os.makedirs('../data/processed', exist_ok=True)

data.to_csv('../data/processed/train.csv', index=False)
test_data.to_csv('../data/processed/test.csv', index=False)
```
