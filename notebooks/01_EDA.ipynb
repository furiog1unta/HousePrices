```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


# показывать таблицы целиком, без «...»
pd.set_option('display.max_columns', 120)
pd.set_option('display.max_rows', 120)
```

```python
data = pd.read_csv('../data/raw/train.csv')
print(data.shape)
data.head()
```

```python
test_data = pd.read_csv('../data/raw/test.csv')
print(test_data.shape)
test_data.head()
```

```python
data.info(max_cols=100)
```

```python
# сколько пропусков в каждой колонке (только там, где они есть)
missing = data.isna().sum()
missing = missing[missing > 0]
missing.sort_values(ascending=False, inplace=True)
sns.barplot(missing, orient='h')
```

```python
# колонки, где пропущено больше 70% значений
miss = pd.DataFrame({'Missing': missing,
                     'Percent': missing / data.shape[0]},
                    index=missing.index)
miss[miss['Percent'] > .7]
```

То же самое для test.

```python
test_data.info()
```

```python
missing_test = test_data.isna().sum()
missing_test = missing_test[missing_test > 0]
missing_test.sort_values(ascending=False, inplace=True)
sns.barplot(missing_test, orient='h')
```

```python
miss_test = pd.DataFrame({'Missing': missing_test,
                     'Percent': missing_test / test_data.shape[0]}, 
                    index=missing_test.index)
miss_test[miss_test['Percent'] > .7].index
```

```python
data.describe()
```

```python
# log1p(x) = log(1 + x): большие цены «сжимаются», хвост становится короче
data['LogSalePrice'] = np.log1p(data['SalePrice'])
data.head()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

sns.histplot(data['SalePrice'], kde=True, ax=ax[0])
ax[0].set_title('Distribution of SalePrice')

sns.histplot(data['LogSalePrice'], kde=True, ax=ax[1])
ax[1].set_title('Distribution of LogSalePrice');
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

sns.boxplot(data['SalePrice'], ax=ax[0], orient='h')
ax[0].set_title('Raw Scale')

sns.boxplot(data['LogSalePrice'], ax=ax[1], orient='h')
ax[1].set_title('Log Scale');
```

```python
# дальше работаем только с логарифмом цены
data.drop(columns=['SalePrice'], inplace=True)
```

```python
# текстовые колонки в pandas имеют тип 'str', все остальные считаем числовыми
numerical_features = data.dtypes[data.dtypes != 'str'].index
print(f'Numerical features: {len(numerical_features)}')

categorical_features = data.dtypes[data.dtypes == 'str'].index
print(f'Categorical featueres: {len(categorical_features)}')
```

```python
data[numerical_features].head()
```

```python
data[categorical_features].head()
```

```python
# признаки от самой сильной связи с ценой к самой слабой
corr = data[numerical_features].corr()
corr['LogSalePrice'].sort_values(ascending=False).index
```

```python
# 10 признаков, сильнее всего связанных с ценой, и сама цена
cols = corr['LogSalePrice'].sort_values(ascending=False).head(11).index
plt.figure(figsize=(12, 10))
correlation = data[cols].corr()
# на карте видно и связь с ценой, и пары признаков, которые почти повторяют друг друга
sns.heatmap(correlation, annot=True, linewidths=0.01, square=True);
```
